```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import os

# Load labels
df = pd.read_excel("data/labels/penalty_labels.xlsx")
print("Dataset loaded successfully!")
print(f"Total penalties: {len(df)}")
print("\n")
print(df)
```

```python
# Chart 1: Shot Direction Distribution
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='shot_direction', palette='viridis')
plt.title('Shot Direction Distribution', fontsize=14)
plt.xlabel('Shot Direction')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='outcome', palette='coolwarm')
plt.title('Penalty Outcome Distribution', fontsize=14)
plt.xlabel('Outcome')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df, x='dominant_foot', palette='Blues')
plt.title('Dominant Foot Distribution', fontsize=14)
plt.xlabel('Dominant Foot')
plt.ylabel('Count')
plt.show()
```

```python
frames_dir = "data/frames"
clip_names = []
frame_counts = []

for folder in sorted(os.listdir(frames_dir)):
    count = len(os.listdir(os.path.join(frames_dir, folder)))
    clip_names.append(folder)
    frame_counts.append(count)

plt.figure(figsize=(10,6))
sns.barplot(x=clip_names, y=frame_counts, palette='magma')
plt.title('Number of Frames Per Penalty Clip', fontsize=14)
plt.xlabel('Penalty Clip')
plt.ylabel('Frame Count')
plt.xticks(rotation=45)
plt.show()
```

```python
keypoints_dir = "data/keypoints"
detection_rates = []

for folder in sorted(os.listdir(keypoints_dir)):
    kp_path = os.path.join(keypoints_dir, folder, "keypoints.npy")
    if os.path.exists(kp_path):
        kp = np.load(kp_path)
        detected = len(kp)
    else:
        detected = 0
    detection_rates.append(detected)

plt.figure(figsize=(10,6))
sns.barplot(x=clip_names, y=detection_rates, palette='rocket')
plt.title('Pose Detection Count Per Penalty Clip', fontsize=14)
plt.xlabel('Penalty Clip')
plt.ylabel('Frames with Pose Detected')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(7,7))
outcome_counts = df['outcome'].value_counts()
plt.pie(outcome_counts, labels=outcome_counts.index, autopct='%1.1f%%',
        colors=['#2ecc71','#e74c3c','#3498db'], startangle=90)
plt.title('Penalty Outcome Breakdown', fontsize=14)
plt.show()
```
